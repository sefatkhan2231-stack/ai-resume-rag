```python
def  chunk_text(text, chunk_size=150, overlap=30):
    words = text.split()

    chunks = []

    start = 0

    while start < len(words):
        end = start + chunk_size

        chunk = " ".join(words[start:end])

        chunks.append(chunk)

        start += chunk_size - overlap

    return chunks
```

```python
def extract_sections(text):
    pattern = r"(?m)^(ABOUT ME|SKILLS|PROFESSIONAL EXPERIENCE|PROJECTS|EDUCATION)\s*$"

    parts = re.split(
        pattern,
        clean_text,
        flags=re.IGNORECASE
    )

    sections = {}

    current_section = None

    for part in parts:
        part = part.strip()

        if not part:
            continue

        section_name = part.upper()

        if section_name in [
            "ABOUT ME",
            "SKILLS",
            "PROFESSIONAL EXPERIENCE",
            "PROJECTS",
            "EDUCATION"
        ]:

            current_section = section_name
            sections[current_section] = ""

        elif current_section:
            sections[current_section] += " " + part

    return sections
```

```python
sections = extract_sections(clean_text)

for section, content in sections.items():
    print("\n" + "=" * 60)
    print(section)
    print("=" * 60)
    print(content)
```

```python
def chunk_section(text, section, max_words=150, overlap=30):
    words = text.split()

    if len(words) <= max_words:
        return [{
            "text": f"{section}\n{text}",
            "section": section
        }]

    chunks = []

    start = 0

    while start < len(words):
        end = start + max_words

        chunk_text = " ".join(words[start:end])

        chunks.append({
            "text": f"{section}\n{chunk_text}",
            "section": section
        })

        start += max_words - overlap

    return chunks
```

```python
all_chunks = []

for section, content in sections.items():

    section_chunks = chunk_section(
        content,
        section,
        max_words=150,
        overlap=30
    )

    all_chunks.extend(section_chunks)

print("Total chunks:", len(all_chunks))
```

```python
def search_resume(query, n_results=3, section=None):

    query_embedding = model.encode(query).tolist()

    query_kwargs = {
        "query_embeddings": [query_embedding],
        "n_results": n_results,
        "include": [
            "documents",
            "metadatas",
            "distances"
        ]
    }

    if section:
        query_kwargs['where'] = {
            "section": section
        }

    results = collection.query(**query_kwargs)

    return results
```

```python
def build_prompt(query, context):

    prompt = f"""
You are an AI resume screening assistant.

Answer the user's question using ONLY the resume
information provided in the context.

Do not invent skills, experience, projects,
education, or qualifications.

If the information is not present in the context,
say that there is not enough evidence.

RESUME CONTEXT:
{context}

USER QUESTION:
{query}

ANSWER:
"""

    return prompt
```

```python
def build_context(results):

    context_parts = []

    for i, document in enumerate(results["documents"][0]):

        metadata = results["metadatas"][0][i]

        context_parts.append(
            f"""
            SOURCE {i + 1}
            Section: {metadata.get("section")}
            Subsection: {metadata.get("subsection", "N/A")}

            {document}
            """
                    )

    return "\n".join(context_parts)
```

```python
def evaluate_retrieval(evaluation_queries):

    correct = 0

    for item in evaluation_queries:

        if item == "What projects has this candidate built?":
            results = search_resume(
                "What projects has this candidate built?",
                n_results=3,
                section="projects"
            )
        else:

            results = search_resume(
                item["query"],
                n_results=1
            )

        metadata = results["metadatas"][0][0]

        predicted_section = metadata.get("section")
        predicted_subsection = metadata.get("subsection")

        expected_section = item["expected_section"]
        expected_subsection = item["expected_subsection"]

        if expected_section is None:
            match = False
        else:
            match = (
                predicted_section == expected_section
                and
                (
                    expected_subsection is None
                    or predicted_subsection == expected_subsection
                )
            )

        if match:
            correct += 1

        print("=" * 70)
        print("QUERY:", item["query"])
        print("EXPECTED:", expected_section, expected_subsection)
        print("PREDICTED:", predicted_section, predicted_subsection)
        print("CORRECT:", match)

    accuracy = correct / len(evaluation_queries)

    print("\nRetrieval Accuracy:", accuracy)
```

```python
def detect_section(query):

    query = query.lower()

    if any(word in query for word in [
        "project",
        "projects",
        "built",
        "application",
        "app"
    ]):
        return "projects"

    if any(word in query for word in [
        "work",
        "worked",
        "experience",
        "job",
        "company",
        "internship"
    ]):
        return "professional_experience"

    if any(word in query for word in [
        "education",
        "study",
        "studied",
        "college",
        "school",
        "degree",
        "training"
    ]):
        return "education"

    if any(word in query for word in [
        "skill",
        "skills",
        "know",
        "experience with"
    ]):
        return "skills"

    return None
```

```python
def rerank_results(query, results):

    documents = results["documents"][0]

    pairs = [
        [query, document]
        for document in documents
    ]

    scores = reranker.predict(pairs)

    ranked = []

    for i, score in enumerate(scores):

        ranked.append({
            "document": documents[i],
            "metadata": results["metadatas"][0][i],
            "distance": results["distances"][0][i],
            "rerank_score": float(score)
        })

    ranked.sort(
        key=lambda x: x["rerank_score"],
        reverse=True
    )

    return ranked
```

```python
def get_top_rerank_score(query):

    results = smart_search(query, n_results=5)
    ranked_results = rerank_results(query, results)

    if not ranked_results:
        return {
            "query": query,
            "score": None,
            "section": None,
            "subsection": None,
            "document": None
        }

    top = ranked_results[0]

    return {
        "query": query,
        "score": top["rerank_score"],
        "section": top["metadata"].get("section"),
        "subsection": top["metadata"].get("subsection"),
        "document": top["document"]
    }
```

```python
THRESHOLD = -10.3


def retrieve_and_rerank(query, n_results=10, max_results=3):

    # 1. Determine relevant section
    section = detect_section(query)

    # 2. Retrieve candidate chunks
    results = search_resume(
        query=query,
        n_results=n_results,
        section=section
    )

    # 3. Rerank
    ranked_results = rerank_results(
        query,
        results
    )

    # 4. Filter weak evidence
    relevant_results = [
        result
        for result in ranked_results
        if result["rerank_score"] >= THRESHOLD
    ]

    # 5. Keep only the strongest evidence
    relevant_results = relevant_results[:max_results]

    return relevant_results
```

```python
query = "Does this candidate know React?"

results = retrieve_and_rerank(query)

for result in results:

    print("=" * 70)
    print("SCORE:", result["rerank_score"])
    print("SECTION:", result["metadata"].get("section"))
    print("SUBSECTION:", result["metadata"].get("subsection"))
    print(result["document"])
```

```python
def generate_answer(query, evidence):

    context_parts = []

    for i, result in enumerate(evidence, start=1):

        section = result["metadata"].get("section", "")
        subsection = result["metadata"].get("subsection", "")

        context_parts.append(
            f"[SOURCE {i}]\n"
            f"Section: {section}\n"
            f"Subsection: {subsection}\n"
            f"Content: {result['document']}"
        )

    context = "\n\n".join(context_parts)

    prompt = f"""
You are an AI resume screening assistant.

Answer the user's question using ONLY the provided resume evidence.

RESUME EVIDENCE:
{context}

USER QUESTION:
{query}

RULES:
1. Use only the provided evidence.
2. Do not invent information.
3. Do not make assumptions.
4. If the evidence does not support an answer, say:
   "No relevant evidence found in the resume."
5. Keep the answer concise.
6. Mention the relevant source section when appropriate.
"""

    response = ollama.chat(
        model="llama3.2",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response["message"]["content"]
```

```python
def ask_resume(query):

    evidence = retrieve_and_rerank(query)

    if not evidence:
        return {
            "answer": "No relevant evidence found in the resume.",
            "sources": []
        }

    answer = generate_answer(
        query,
        evidence
    )

    sources = []

    for result in evidence:

        sources.append({
            "section": result["metadata"].get("section"),
            "subsection": result["metadata"].get("subsection"),
            "score": result["rerank_score"],
            "document": result["document"]
        })

    return {
        "answer": answer,
        "sources": sources
    }
```

```python
import re


SKILL_NORMALIZATION = {
    "Python programming": "Python",
    "SQL/database knowledge": "SQL",
    "Experience working with Git and GitHub": "Git GitHub",
    "Good problem solving skills": "problem solving",
}


def normalize_skill(skill):
    skill = skill.strip()

    return SKILL_NORMALIZATION.get(
        skill,
        skill
    )


def verify_evidence(skill, evidence):

    skill_terms = re.findall(
        r"[a-z0-9+#.-]+",
        skill.lower()
    )

    for result in evidence:

        document = result.get("document", "").lower()

        if all(
            re.search(
                rf"\b{re.escape(term)}\b",
                document
            )
            for term in skill_terms
        ):
            return True

    return False


def check_skill(skill):

    # 1. Normalize the requirement
    normalized_skill = normalize_skill(skill)

    # 2. Create the retrieval query
    query = f"""
Is {normalized_skill} explicitly mentioned
or supported by evidence in this candidate's resume?
"""

    # 3. Retrieve + rerank possible evidence
    evidence = retrieve_and_rerank(query)

    # 4. Nothing retrieved = no evidence
    if not evidence:
        return {
            "skill": skill,
            "normalized_skill": normalized_skill,
            "matched": False,
            "evidence": []
        }

    # 5. IMPORTANT:
    verified = verify_evidence(
        normalized_skill,
        evidence
    )

    if not verified:
        return {
            "skill": skill,
            "normalized_skill": normalized_skill,
            "matched": False,
            "evidence": []
        }

    # 6. Return the verified result
    return {
        "skill": skill,
        "normalized_skill": normalized_skill,
        "matched": verified,
        "evidence": evidence
    }
```

```python
def analyze_requirements(requirements):

    matched = []
    missing = []

    for skill in requirements:

        result = check_skill(skill)

        if result["matched"]:
            matched.append(result)

        else:
            missing.append(result)

    return {
        "matched": matched,
        "missing": missing
    }
```

```python
REQUIREMENT_WEIGHTS = {
    "Python": 1.0,
    "machine learning algorithms": 1.0,
    "NLP": 1.0,
    "Transformers": 1.0,
    "scikit-learn": 0.8,
    "PyTorch": 1.0,
    "TensorFlow": 1.0,
    "SQL": 0.8,
    "Git": 0.4,
    "GitHub": 0.4,
}
```

```python
def calculate_match_score(results):

    total_weight = 0
    matched_weight = 0

    for result in results:

        skill = result['skill']

        weight = REQUIREMENT_WEIGHTS.get(
            skill,
            1.0
        )

        total_weight += weight

        if result["matched"]:
            matched_weight += weight

    if total_weight == 0:
        return 0

    return (
        matched_weight / total_weight
    ) * 100
```

```python
def analyze_requirements(requirements):

    results = []

    for skill in requirements:

        result = check_skill(skill)

        results.append(result)

    score = calculate_match_score(results)

    return {
        "score": score,
        "results": results
    }
```

```python
def build_evidence_context(analysis):

    grouped = {}

    for result in analysis["results"]:

        if not result["matched"]:
            continue

        skill = result["skill"]

        if skill not in grouped:
            grouped[skill] = []

        for evidence in result["evidence"]:

            metadata = evidence["metadata"]

            section = metadata.get("section")
            subsection = metadata.get("subsection")
            document = evidence["document"]

            evidence_item = (
                section,
                subsection,
                document
            )

            if evidence_item not in grouped[skill]:
                grouped[skill].append(evidence_item)

    context_parts = []

    for skill, evidence_list in grouped.items():

        block = f"Requirement: {skill}\n"

        for section, subsection, document in evidence_list:

            block += (
                f"\nSource: {section} → {subsection}\n"
                f"Evidence:\n{document}\n"
            )

        context_parts.append(block)

    return "\n".join(context_parts)
```

```python
def build_candidate_prompt(analysis):

    evidence_context = build_evidence_context(
        analysis
    )

    missing = [
        result["skill"]
        for result in analysis["results"]
        if not result["matched"]
    ]

    prompt = f"""
You are an AI resume screening assistant.

Analyze the candidate based ONLY on the
verified resume evidence provided below.

Do not invent skills or experience.

MATCH SCORE:
{analysis['score']:.2f}%

VERIFIED EVIDENCE:
{evidence_context}

MISSING REQUIREMENTS:
{chr(10).join("- " + skill for skill in missing)}

Write a concise candidate assessment.

The candidate's verified match score is the percentage
of evaluated requirements with supporting resume evidence.

Do not describe a partial match as having "no alignment".

Distinguish between:
- confirmed matches
- missing requirements
- overall match level

Do not make a hiring decision unless an explicit hiring
threshold is provided.

Include:

1. Overall assessment
2. Confirmed strengths
3. Missing requirements
4. Final recommendation

Important rules:

- Only use the evidence provided.
- Do not assume a skill from a related technology.
- Python does not imply machine learning.
- Python does not imply scikit-learn.
- Transformers do not imply NLP experience.
- If evidence is missing, explicitly say so.
"""

    return prompt
```

```python
def verify_skill(skill, evidence):

    prompt = f"""
You are checking whether a resume contains evidence for a requirement.

REQUIREMENT:
{skill}

RESUME EVIDENCE:
{evidence}

Does the evidence support the requirement?

For this task:

- If the exact skill appears in the evidence, answer TRUE.
- If the evidence clearly names the required technology, answer TRUE.
- Otherwise answer FALSE.

Examples:

Requirement: SQL
Evidence: JavaScript, Python, SQL, HTML
TRUE

Requirement: Python
Evidence: JavaScript, Python, PHP
TRUE

Requirement: GitHub
Evidence: Git, GitHub
TRUE

Requirement: PyTorch
Evidence: Python, Git, GitHub
FALSE

Return ONLY:
TRUE
or
FALSE
"""

    response = ollama.chat(
        model=MODEL_NAME,
        messages=[
            {
                "role": "system",
                "content": "Return only TRUE or FALSE."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        options={
            "temperature": 0
        }
    )

    answer = response.message.content.strip().upper()

    print("VERIFIER:", skill, "→", repr(answer))

    return answer == "TRUE"
```

```python
import re


def normalize_text(text):
    text = text.lower()
    text = re.sub(r"[^a-z0-9+#.\- ]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()


def exact_skill_match(skill, evidence):

    skill = normalize_text(skill)
    evidence = normalize_text(evidence)

    return skill in evidence
```
