```python
from pypdf import PdfReader
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
pdf_path = "../dataset/sefat_khan.pdf"

reader = PdfReader(pdf_path)

text = ""

for page in reader.pages:
    page_text = page.extract_text()

    if page_text:
        text += page_text + "\n"

print(text[:3000])
```

```python
print(type(text))
print(len(text))
```

```python
words = text.split()

chunk_size = 100

chunks = []

for i in range(0, len(words), chunk_size):
    chunk = " ".join(words[i:i + chunk_size])
    chunks.append(chunk)

print("Number of chunks:", len(chunks))
```

```python
for i, chunk in enumerate(chunks):
    print(f"\n--- Chunk {i} ---")
    print(chunk[:500])
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
print(model)
```

```python
sentence = "I am a Python developer with machine learning experience."

embedding = model.encode(sentence)

print(type(embedding))
print(embedding.shape)
```

```python
sentences = [
    "I am a Python developer with machine learning experience.",
    "I develop machine learning applications using Python.",
    "I enjoy playing football and watching cricket."
]
```

```python
embeddings = model.encode(sentences)

print(embeddings.shape)
```

```python
print(embedding[:10])
```

```python
similarity_01 = cosine_similarity(
    [embeddings[0]],
    [embeddings[1]]
)

similarity_02 = cosine_similarity(
    [embeddings[0]],
    [embeddings[2]]
)

print("Sentence 1 vs Sentence 2:", similarity_01[0][0])
print("Sentence 1 vs Sentence 3:", similarity_02[0][0])
```

```python
sentences = [
    "I am an experienced Python developer.",
    "I have professional experience developing software with Python.",
    "I am a frontend React developer.",
    "I enjoy playing football."
]

embeddings = model.encode(sentences)

for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        score = cosine_similarity(
            [embeddings[i]],
            [embeddings[j]]
        )[0][0]

        print(f"{i} vs {j}: {score:.4f}")
        print("  ", sentences[i])
        print("  ", sentences[j])
        print()
```

```python
job_description = """
We are looking for a Machine Learning Engineer
with strong Python programming skills,
experience with NLP and Transformers,
and knowledge of SQL.
"""
```

```python
resume_chunks = [
    "Developed machine learning models using Python and scikit-learn.",
    "Built NLP classification systems using BERT and Transformers.",
    "Developed React applications using JavaScript.",
    "Worked with MySQL and SQL databases.",
    "Played football for the university team."
]
```

```python
job_embedding = model.encode([job_description])

chunk_embeddings = model.encode(resume_chunks)
```

```python
scores = cosine_similarity(
    job_embedding,
    chunk_embeddings
)[0]
```

```python
for chunk, score in zip(resume_chunks, scores):
    print(f"{score:.4f} → {chunk}")
```

```python
queries = [
    "Python programming experience",
    "Natural language processing and Transformers experience",
    "SQL database experience",
]
```

```python
resume_chunks = [
    "Developed machine learning models using Python and scikit-learn.",
    "Built NLP classification systems using BERT and Transformers.",
    "Developed React applications using JavaScript.",
    "Worked with MySQL and SQL databases.",
    "Played football for the university team."
]
```

```python
query_embeddings = model.encode(queries)
chunk_embeddings = model.encode(resume_chunks)

similarities = cosine_similarity(
    query_embeddings,
    chunk_embeddings
)

for i, query in enumerate(queries):
    print(f"\nQUERY: {query}")
    
    for j, chunk in enumerate(resume_chunks):
        print(f"{similarities[i][j]:.4f} → {chunk}")
```
